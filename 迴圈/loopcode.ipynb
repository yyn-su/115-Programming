# loop
loop指令可讓程式碼不依條件重複執行，直到滿足條件為止。

```python
# 底下程式碼會印出0-9
for i in range(10):
    print(i)
```

```python
list(range(10))
```

```python
# 底下程式碼會印出0-9
for i in range(10):
    print(i+1)
```

```python
sum = 0
for i in range(10):
    sum = sum + (i+1)
print(sum)
```

# range

- 懶惰是一種美德
- 由0開頭
- 包含頭不包含尾

```python
print(list(range(5)))
# 區間由0開始到5-1結束
print(list(range(0,5)))
# 區間由1開始到5-1結束
print(list(range(1,5)))
# 區間由1開始到5-1結束 每次跳兩格
print(list(range(1,5,2)))
```

```python
print(list(range(2,101,2)))
```
