# 簡介
串烈是一個可以儲存多個元素(變數、浮點數、字串)的容器。
串列中每個元素都是用逗點隔開，左右有[]


```python
# score 是一個串列，裡面每個元素都是整數
scores = [78,29,56,80,99]
print(scores)
```

# 索引值
我們可以使用索引值(index)來存取列中的元素。
注意:索引值由0開始。

```python
print(scores[0]) # 印出第一個元素
print(scores[1]) # 印出第二個元素
print(scores[2]) # 印出第三個元素
```

## 串列內容的個數
list 是 length(長度)的意思

```python
print(len(scores))
print(scores[4]) # 取得最後一個元素
print(scores[5-1]) # 取得最後一個元素
print(scores[-1]) # 取得最後一個元素
print(scores[len(scores)-1])
```

## 新增元素
使用`append()`新增元素到串列的最尾端

```python
print(scores)
scores[2] = 70 # 將70指定給索引值為2的元素
print(scores)
```

```python
print(scores)
scores.append(78)
scores.append(93)
print(scores)
```

```python
weights = [12]*6
print(weights)
```

## 串列三種寫法

```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```

```python
# 對於 scores 中的每個元素，一次取一個放在 e 中     最推薦的寫法
for e in scores:
    print(e)
```

```python
for i,e in enumerate(scores):
    print(i,e)
```

# 計算賈伯納西數列的第91個元素
賈伯納西數列:
1,1,2,3,5,8...
91元素為?
提示:
設 $x$ 是數列，


```python
# 初始化串列，包和91個0
fib = [0]*91
print(fib)
fib[0]=1
fib[1]=1
print(fib)
fib[2]=fib[0]+fib[1]
print(fib)
fib[3]=fib[1]+fib[2]
print(fib)
fib[4]=fib[2]+fib[3]
print(fib)
```

```python
for i in range(2,91):
    fib[i] = fib[i-2] + fib[i-1]

print(fib[90])
```
